```python
data_v1=pd.read_csv('data/Train-test-dataset_Ver1.csv')
data_v3=pd.read_csv('data/Train-test-dataset_Ver3.csv')
```

```python
import matplotlib.pyplot as plt
# 计算均值和最大值
mean_v1 = data_v1.mean()
max_v1 = data_v1.max()
mean_v3 = data_v3.mean()
max_v3 = data_v3.max()

# 创建一个DataFrame来比较均值和最大值
comparison = pd.DataFrame({
    'Mean V1': mean_v1,
    'Mean V3': mean_v3,
    'Max V1': max_v1,
    'Max V3': max_v3
})

# 显示比较结果
print("Comparison of Means and Maximums:")
print(comparison)

# 可视化均值
plt.figure(figsize=(12, 6))
plt.plot(mean_v1, label='Mean V1', marker='o')
plt.plot(mean_v3, label='Mean V3', marker='o')
plt.title('Comparison of Means')
plt.xlabel('Columns')
plt.ylabel('Mean Value')
plt.legend()
plt.xticks(rotation=90)  # 旋转x轴标签以便更好地显示
plt.tight_layout()
plt.show()

# 可视化最大值
plt.figure(figsize=(12, 6))
plt.plot(max_v1, label='Max V1', marker='o')
plt.plot(max_v3, label='Max V3', marker='o')
plt.title('Comparison of Maximums')
plt.xlabel('Columns')
plt.ylabel('Maximum Value')
plt.legend()
plt.xticks(rotation=90)  # 旋转x轴标签以便更好地显示
plt.tight_layout()
plt.show()
```
