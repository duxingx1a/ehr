```python
import ehr_utils
import ehr_models
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score, accuracy_score

# 加载数据
file_path_mean = 'data_cleaned/ver-noelectrolyte-logtransform_mean.csv'
data = pd.read_csv(file_path_mean)
X = data.drop(columns=['have_stone'])  # 假设目标列名为 'target_column'
Y = data['have_stone']

# 初始化模型
model = ehr_models.init_XGBoost(False)  # 假设模型已经加载

# 使用 10 折交叉验证
kf = KFold(n_splits=10, shuffle=True, random_state=42)
fold_results = []

for fold, (train_index, test_index) in enumerate(kf.split(X)):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    Y_train, Y_test = Y.iloc[train_index], Y.iloc[test_index]
    
    # 训练模型
    model.fit(X_train, Y_train)
    
    # 预测
    Y_prob = model.predict_proba(X_test)[:, 1]  # 获取概率
    Y_pred = model.predict(X_test)  # 获取预测结果
    
    # 评估模型
    roc_auc = roc_auc_score(Y_test, Y_prob)
    precision = precision_score(Y_test, Y_pred)
    recall = recall_score(Y_test, Y_pred)
    f1 = f1_score(Y_test, Y_pred)
    accuracy = accuracy_score(Y_test, Y_pred)
    
    fold_results.append({
        'fold': fold + 1,
        'roc_auc': roc_auc,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'accuracy': accuracy
    })

# 打印每一折的结果
for result in fold_results:
    print(f"Fold {result['fold']}:")
    print(f"ROC AUC: {result['roc_auc']:.4f}")
    print(f"Precision: {result['precision']:.4f}")
    print(f"Recall: {result['recall']:.4f}")
    print(f"F1 Score: {result['f1_score']:.4f}")
    print(f"Accuracy: {result['accuracy']:.4f}")
    print("-" * 30)

# 计算平均性能
avg_results = {
    'roc_auc': sum([r['roc_auc'] for r in fold_results]) / len(fold_results),
    'precision': sum([r['precision'] for r in fold_results]) / len(fold_results),
    'recall': sum([r['recall'] for r in fold_results]) / len(fold_results),
    'f1_score': sum([r['f1_score'] for r in fold_results]) / len(fold_results),
    'accuracy': sum([r['accuracy'] for r in fold_results]) / len(fold_results)
}

print("Average Performance Across Folds:")
print(f"ROC AUC: {avg_results['roc_auc']:.4f}")
print(f"Precision: {avg_results['precision']:.4f}")
print(f"Recall: {avg_results['recall']:.4f}")
print(f"F1 Score: {avg_results['f1_score']:.4f}")
print(f"Accuracy: {avg_results['accuracy']:.4f}")
```
