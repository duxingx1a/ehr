```python
import pandas as pd
import serach_utils
data_mean=pd.read_csv('../data_cleaned/ver-noelectrolyte-logtransform_mean.csv')
X_train,X_test,Y_train,Y_test=serach_utils.get_train_test(data_mean)
```

```python
from sklearn.ensemble import RandomForestClassifier
from skopt.space import Integer

param_space_randomForest = {
    'n_estimators': Integer(50, 200),  # 基估计器的数量
    'max_depth': Integer(3, 20),  # 最大深度
    'min_samples_split': Integer(2, 10),  # 内部节点再划分所需最小样本数
    'min_samples_leaf': Integer(1, 5)  # 叶子节点最少样本数
}

# 初始化RandomForest分类器
model = RandomForestClassifier(n_jobs=-1, random_state=42,class_weight={0: 1, 1: 9})
serach_utils.search_params(param_space_randomForest, model, X_train, Y_train)
```
