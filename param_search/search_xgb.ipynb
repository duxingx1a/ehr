```python
import pandas as pd
import serach_utils
data_mean=pd.read_csv('../data_cleaned/ver-noelectrolyte-crystal_removed_history_removed_mean.csv')
X_train,X_test,Y_train,Y_test=serach_utils.get_train_test(data_mean)
# col = [
#     'Gender', 'stone_history', 'Uric_WBC', 'Uric_RBC', 'Age', 'Uric_conductivity', 'Alcohol', 'Uric_bacteria', 'Hypertension', 'Systolic_BP', 'Uric_specific_gravity', 'Diabete',
#     'Uric_epithelium', 'Smoke', 'Uric_PH', 'α-HBDH', 'Serum_Uric_acid', 'Creatinine', 'Basophil_percentage', 'Globulin'
# ]

# X_train = X_train[col]
# X_test = X_test[col]
```

```python
from xgboost import XGBClassifier
from skopt.space import Integer, Real

param_space = {
    'n_estimators': Integer(50, 500),  # 迭代次数
    'max_depth': Integer(1, 10),  # 树的最大深度
    'subsample': Real(0.5, 1.0),  # 子采样比例
    'colsample_bytree': Real(0.5, 1.0),  # 列采样比例
    'learning_rate': Real(0.01, 0.3),  # 学习率
    'gamma': Real(0, 10),  # 惩罚项中叶子结点个数前的参数
    'min_child_weight': Real(0, 10),  # 叶子节点最小权重
    'reg_alpha': Real(0, 1),  # L1正则化项的参数
    'reg_lambda': Real(1, 10),  # L2正则化项的参数
    'scale_pos_weight': Real(1, 10)  # 类别不平衡时的权重调整
}

# 初始化XGBoost分类器
model = XGBClassifier(device='gpu', eval_metric='auc',scale_pos_weight=30.0)
serach_utils.search_params(param_space, model, X_train, Y_train)
```
